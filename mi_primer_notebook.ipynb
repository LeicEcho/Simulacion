### Tipado Dinámico: Declarar Variables sin Especificar su Tipo

¡Sí, en Python **es totalmente cierto** que no necesitas especificar el tipo de dato! Python utiliza **tipado dinámico**. 

- **Inferencia de tipo**: El intérprete de Python averigua automáticamente el tipo de dato en el momento en que le asignas un valor con el signo `=`. 
- **Reasignación libre**: Una variable puede cambiar de tipo si más adelante le asignas un valor diferente (por ejemplo, pasar de guardar un número a guardar texto). En lenguajes como Java o C++, esto arrojaría un error.

```python
# Celda 4: Demostración de Tipado Dinámico

# Declaramos la variable 'mi_variable' asignándole un número entero
mi_variable = 100
print(f"Valor: {mi_variable} | Tipo de dato: {type(mi_variable)}")

# Asignamos un número decimal (flotante) a la misma variable
mi_variable = 19.99
print(f"Valor: {mi_variable} | Tipo de dato: {type(mi_variable)}")

# Ahora le asignamos texto (string)
mi_variable = "¡Hola, ahora soy texto!"
print(f"Valor: {mi_variable} | Tipo de dato: {type(mi_variable)}")

# Por último, le asignamos un valor booleano (True/False)
mi_variable = True
print(f"Valor: {mi_variable} | Tipo de dato: {type(mi_variable)}")
```

### Entrada de Datos: Pedir información al usuario e imprimirla

Para recibir datos del usuario desde el teclado, Python nos provee la función `input()`. 

> [!IMPORTANT]
> La función `input()` **siempre devuelve el dato como una cadena de texto (str)**. Si necesitas trabajar con números enteros (`int`) o decimales (`float`), debes convertirlos explícitamente (hacer un "cast").

```python
# Celda 5: Entrada de texto, enteros y flotantes

# 1. Entrada de texto (String)
nombre = input("Escribe tu nombre: ")
print(f"-> Hola {nombre}! Tipo de dato recibido: {type(nombre)}")

# 2. Entrada de un número entero (Integer) - requiere conversión int()
edad_texto = input("Escribe tu edad (entero): ")
edad = int(edad_texto)
print(f"-> Edad: {edad} años | Tipo de dato convertido: {type(edad)}")

# 3. Entrada de un número decimal (Float) - requiere conversión float()
estatura_texto = input("Escribe tu estatura en metros (ej. 1.75): ")
estatura = float(estatura_texto)
print(f"-> Estatura: {estatura}m | Tipo de dato convertido: {type(estatura)}")
```
