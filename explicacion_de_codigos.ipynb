## 1. Explicación de `hola_mundo.py`

Este es el script de inicio clásico. Introduce la estructura de funciones y el punto de entrada de los programas de Python.

### Explicación bloque por bloque:

1. **Comentarios (`# ...`)**:
   ```python
   # Este es un script de Python clásico (.py)
   ```
   Cualquier línea que comience con el símbolo `#` es ignorada por Python. Se utiliza para documentar y explicar qué hace el código.

2. **Definición de Función (`def saludar(...)`):**
   ```python
   def saludar(nombre="Mundo"):
   ```
   - `def` le indica a Python que estamos creando (definiendo) una función llamada `saludar`.
   - `(nombre="Mundo")` es un **parámetro con valor por defecto**. Si llamamos a la función sin pasarle un nombre, tomará automáticamente el valor `"Mundo"`.
   - Los dos puntos `:` al final indican que empieza el bloque de código de la función. Todo lo que esté **indentado** (con espacio hacia la derecha) debajo de esta línea pertenece a la función.

3. **El cuerpo de la función (`print`)**:
   ```python
       print(f"¡Hola, {nombre}!")
       print("El entorno de desarrollo de Python está funcionando correctamente.")
   ```
   - `print()` escribe texto en la consola.
   - La `f` antes de las comillas en `f"¡Hola, {nombre}!"` indica que es un **f-string** (texto formateado). Permite insertar variables directamente dentro del texto rodeándolas con llaves `{}`.

4. **Bloque de Entrada Principal (`__main__`)**:
   ```python
   if __name__ == "__main__":
       saludar()
   ```
   Esta es una buena práctica de Python. Le dice al intérprete: *"Si este archivo se está ejecutando directamente (y no siendo importado por otro archivo), entonces ejecuta la función `saludar()`"*.

````python
# Código de 'hola_mundo.py' listo para ejecutar en el Notebook:

def saludar(nombre="Mundo"):
    print(f"¡Hola, {nombre}!")
    print("El entorno de desarrollo de Python está funcionando correctamente.")

if __name__ == "__main__":
    # Probamos llamarla sin parámetros (usa el valor por defecto)
    saludar()
    print("---")
    # Probamos llamarla con un nombre personalizado
    saludar("Programador")
````

## 2. Explicación de `suma.py`

Este script enseña cómo las funciones devuelven valores utilizando `return` y introduce el manejo básico de errores.

### Explicación bloque por bloque:

1. **Retorno de Funciones (`return`):**
   ```python
   def sumar_numeros(a, b):
       return a + b
   ```
   La palabra clave `return` le dice a la función que calcule la operación `a + b` y **devuelva** el resultado al lugar donde fue llamada la función, para poder guardarlo en una variable.

2. **Ejecución y Captura:**
   ```python
   resultado = sumar_numeros(num1, num2)
   ```
   Llamamos a la función pasándole `num1` y `num2`, y el valor devuelto se almacena en la variable `resultado`.

3. **Control de Errores (`try` y `except`):**
   ```python
   try:
       entrada1 = float(input("Escribe el primer número: "))
       ...
   except ValueError:
       print("Error: Por favor, introduce solo números válidos.")
   ```
   - Si un usuario ingresa letras en lugar de números, `float(input(...))` fallará y arrojará un error de tipo `ValueError`.
   - El bloque `try` (intentar) ejecuta el código normalmente. Si ocurre un error de tipo `ValueError`, el programa **no se cierra de golpe**, sino que salta inmediatamente al bloque `except` (excepción) e imprime un mensaje amigable.

````python
# Código de 'suma.py' listo para ejecutar:

def sumar_numeros(a, b):
    return a + b

if __name__ == "__main__":
    num1 = 12.5
    num2 = 7.3
    resultado = sumar_numeros(num1, num2)
    print(f"Suma de prueba: {num1} + {num2} = {resultado}\n")
    
    # Prueba a ingresar letras para ver cómo actúa el bloque 'except'
    try:
        entrada1 = float(input("Introduce un número: "))
        entrada2 = float(input("Introduce otro número: "))
        print(f"Suma interactiva: {sumar_numeros(entrada1, entrada2)}")
    except ValueError:
        print("\n[!] Capturamos un error: No ingresaste un número válido, pero el programa continuó a salvo.")
````

## 3. Explicación de `tipado_dinamico.py`

Este archivo ilustra el comportamiento del tipado dinámico en Python, donde las variables cambian de tipo según su contenido.

### Explicación bloque por bloque:

1. **Reasignación Libre:**
   ```python
   variable = 42
   ...
   variable = "Texto en la misma variable"
   ```
   Declaramos una variable llamada `variable`. Primero le asignamos un entero, y luego, sin necesidad de destruir la variable o declarar una nueva, le asignamos una cadena de texto. Python gestiona la memoria y el cambio de tipo de forma automática por debajo.

2. **La función `type()`:**
   ```python
   type(variable)
   ```
   Esta es una función incorporada en Python que te devuelve la clase (el tipo de dato) del objeto que le pases como parámetro. Por ejemplo, te dirá `<class 'int'>` para enteros o `<class 'str'>` para texto.

````python
# Código de 'tipado_dinamico.py' listo para ejecutar:

def demostrar_tipado_dinamico():
    # Guardamos un entero
    variable = 42
    print(f"Valor: {variable} | Tipo: {type(variable)}")
    
    # Cambiamos a texto
    variable = "Texto en la misma variable"
    print(f"Valor: '{variable}' | Tipo: {type(variable)}")
    
    # Cambiamos a una lista
    variable = [1, 2, 3]
    print(f"Valor: {variable} | Tipo: {type(variable)}")

demostrar_tipado_dinamico()
````

## 4. Explicación de `entrada_datos.py`

Muestra detalladamente cómo leer datos externos, la diferencia de tipos al recibirlos, y cómo estructurar conversiones seguras.

### Explicación bloque por bloque:

1. **Entrada de Texto por Defecto:**
   ```python
   nombre = input("1. Introduce tu nombre: ")
   ```
   `input()` detiene la ejecución del programa y espera que el usuario escriba algo y presione `Enter`. El dato siempre ingresa como `<class 'str'>`.

2. **Casteo a Entero (`int()`):**
   ```python
   edad = int(input("..."))
   ```
   Toma la cadena de texto numérica (por ejemplo, `"25"`) y la transforma en el número entero (`25`), permitiendo hacer operaciones matemáticas (como sumas o comparaciones) con él.

3. **Casteo a Flotante (`float()`):**
   ```python
   estatura = float(input("..."))
   ```
   Funciona igual que `int()`, pero convierte el texto en un número con decimales (por ejemplo, `"1.75"` a `1.75`). Recuerda que en Python los decimales usan punto `.`, no coma `,`.

````python
# Código de 'entrada_datos.py' listo para ejecutar:

def capturar_datos():
    # 1. Texto
    nombre = input("1. Escribe tu nombre (Texto): ")
    print(f"-> Guardado: '{nombre}' | Tipo: {type(nombre)}\n")
    
    # 2. Entero con conversión
    try:
        edad = int(input("2. Escribe tu edad (Entero): "))
        print(f"-> Guardado: {edad} | Tipo: {type(edad)}\n")
    except ValueError:
        print("[!] No pudimos convertir a entero. Ingresaste texto inválido.\n")
        
    # 3. Flotante con conversión
    try:
        estatura = float(input("3. Escribe tu estatura (Decimal con punto): "))
        print(f"-> Guardado: {estatura} | Tipo: {type(estatura)}\n")
    except ValueError:
        print("[!] No pudimos convertir a flotante. Ingresaste texto inválido.\n")

capturar_datos()
````
