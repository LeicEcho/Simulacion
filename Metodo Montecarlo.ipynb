# Método de Montecarlo: Fundamentos, Simulación y Aplicaciones

El **Método de Montecarlo** es una de las herramientas numéricas y computacionales más potentes y versátiles de la actualidad. Utilizado para aproximar soluciones de problemas matemáticos complejos y simular sistemas con incertidumbre, su principio fundamental radica en el uso de **muestreos aleatorios repetidos** para obtener resultados numéricos aproximados.

Este cuaderno ofrece una revisión teórica rigurosa del método, cubriendo desde sus orígenes históricos y sustento matemático, hasta sus aplicaciones en la industria.

---

## 1. Introducción y Origen Histórico

### Definición
El Método de Montecarlo es un enfoque numérico no determinista que resuelve problemas deterministas o probabilísticos a través de la simulación de variables aleatorias. En lugar de resolver analíticamente ecuaciones matemáticas complejas (lo cual a menudo es imposible), se modelan las entradas aleatorias del sistema y se simula su comportamiento miles o millones de veces para estimar la distribución de probabilidad de las salidas.

### Origen Histórico
El nacimiento formal del método se sitúa en la década de **1940** en el **Laboratorio Nacional de Los Álamos** (Nuevo México, EE. UU.), en el seno del **Proyecto Manhattan** para el desarrollo de la bomba atómica.

1. **La idea inicial (Stanislaw Ulam)**: Mientras se recuperaba de una enfermedad en 1946 jugando al solitario, el matemático polaco-estadounidense Stanislaw Ulam se preguntó cuál era la probabilidad de ganar una partida de solitario determinada. Al notar que calcular la probabilidad exacta de forma analítica requería un análisis combinatorio intratable, pensó en una alternativa práctica: jugar el solitario 100 veces, contar las victorias y usar esa proporción como una estimación. Con la reciente creación de la computadora **ENIAC**, Ulam comprendió que esta idea de experimentación estadística podía automatizarse a gran velocidad.
2. **El desarrollo teórico (John von Neumann)**: Ulam compartió la idea con John von Neumann, quien vio de inmediato la aplicación del método para resolver ecuaciones diferenciales complejas sobre la difusión y comportamiento de neutrones en materiales de fisión (un problema crítico en Los Álamos donde los experimentos físicos reales eran peligrosos y costosos). Von Neumann desarrolló los algoritmos matemáticos necesarios, incluyendo algunos de los primeros generadores de números pseudoaleatorios.
3. **El nombre en clave (Nicholas Metropolis)**: El método requería un nombre clave secreto debido al carácter clasificado del proyecto. El físico Nicholas Metropolis propuso el nombre **"Montecarlo"**, inspirado en el famoso Casino de Montecarlo en el Principado de Mónaco. Esto se debió a que el tío de Ulam solía pedir dinero prestado para apostar en este casino, y los juegos de azar (como la ruleta, los dados y las cartas) representan la esencia misma de la aleatoriedad y probabilidad en las que se basa el método.



## 2. Conceptos Teóricos Fundamentales

El Método de Montecarlo no es una simple técnica heurística, sino que está firmemente respaldado por la teoría matemática de la probabilidad y la estadística.

### 2.1 La Ley de los Grandes Números (LGN)
La Ley de los Grandes Números es el pilar matemático que garantiza que las estimaciones de Montecarlo convergen al valor real a medida que aumenta el número de simulaciones.

Supongamos que $X_1, X_2, \dots, X_N$ son variables aleatorias independientes e idénticamente distribuidas (i.i.d.) con un valor esperado común $\mathbb{E}[X_i] = \mu$ y una varianza $\sigma^2 < \infty$. La media muestral se define como:
$$\bar{X}_N = \frac{1}{N} \sum_{i=1}^{N} X_i$$

* **Ley Débil de los Grandes Números**: Establece que la media muestral converge en probabilidad hacia la media poblacional $\mu$:
$$\lim_{N \to \infty} P\left( \left| \bar{X}_N - \mu \right| < \epsilon \right) = 1 \quad \forall \epsilon > 0$$

* **Ley Fuerte de los Grandes Números**: Establece que la media muestral converge casi seguramente (con probabilidad 1) a $\mu$:
$$P\left( \lim_{N \to \infty} \bar{X}_N = \mu \right) = 1$$

En el contexto de Montecarlo, esto significa que si simulamos un proceso un número suficientemente grande de veces ($N \to \infty$), el promedio de nuestros resultados experimentales se acercará tanto como queramos al valor esperado teórico.

---

### 2.2 Estimación de Integrales por Montecarlo
Una de las aplicaciones matemáticas más directas del método es la integración numérica, especialmente útil en dimensiones altas donde los métodos de cuadratura clásicos (como la regla del trapecio o de Simpson) sufren de la "maldición de la dimensionalidad".

Queremos calcular la integral de una función $f(x)$ en el intervalo $[a, b]$:
$$I = \int_{a}^{b} f(x) \, dx$$

El valor promedio de la función $f(x)$ sobre $[a, b]$ está definido como:
$$\mathbb{E}[f(X)] = \int_{a}^{b} f(x) \cdot g(x) \, dx$$
donde $g(x) = \frac{1}{b-a}$ es la función de densidad de probabilidad de una variable aleatoria uniforme en el intervalo $[a, b]$, denotada como $X \sim \mathcal{U}(a, b)$.

Reescribiendo la integral original:
$$I = (b-a) \int_{a}^{b} f(x) \frac{1}{b-a} \, dx = (b-a) \mathbb{E}[f(X)]$$

Aplicando la Ley de los Grandes Números, podemos aproximar el valor esperado $\mathbb{E}[f(X)]$ generando $N$ números aleatorios independientes $x_1, x_2, \dots, x_N$ distribuidos uniformemente en $[a, b]$ y calculando el promedio de la función evaluada en estos puntos:
$$\hat{I}_N = (b-a) \cdot \frac{1}{N} \sum_{i=1}^{N} f(x_i)$$

#### Error y Convergencia
De acuerdo con el **Teorema del Límite Central (TLC)**, para un $N$ grande, el error de estimación está distribuido aproximadamente de forma normal con media 0 y varianza:
$$\text{Var}(\hat{I}_N) = \frac{(b-a)^2}{N} \text{Var}(f(X))$$

Por lo tanto, la desviación estándar del error (el error estándar) decrece a una tasa de:
$$\text{Error} \propto \mathcal{O}\left(\frac{1}{\sqrt{N}}\right)$$

Esta tasa de convergencia de $\mathcal{O}(N^{-1/2})$ es independiente del número de dimensiones de la integral, lo que explica por qué Montecarlo es el método preferido para integrales multidimensionales (por ejemplo, en física estadística o finanzas cuantitativas).

### 2.3 Generación de Números Pseudoaleatorios (PRNG)
Para realizar simulaciones en una computadora, necesitamos una fuente de aleatoriedad. Dado que las computadoras son máquinas deterministas, generan **números pseudoaleatorios**: secuencias numéricas calculadas mediante algoritmos matemáticos que aparentan ser aleatorias al pasar diversas pruebas estadísticas.

#### Generador Congruencial Lineal (LCG)
Es uno de los algoritmos históricos más sencillos y conocidos. Está definido por la relación de recurrencia:
$$X_{n+1} = (a \cdot X_n + c) \pmod m$$
Donde:
* $X_0$ es la **semilla** (seed), el valor inicial que determina toda la secuencia.
* $a$ es el **multiplicador**.
* $c$ es el **incremento**.
* $m$ es el **módulo** ($m > 0$).

Para obtener un número pseudoaleatorio distribuido uniformemente en el intervalo $[0, 1)$, se calcula:
$$U_n = \frac{X_n}{m}$$

#### Generación de Distribuciones No Uniformes
A partir de una variable uniforme $U \sim \mathcal{U}(0, 1)$, podemos generar variables aleatorias que sigan otras distribuciones de probabilidad mediante diversas técnicas:

1. **Método de la Transformada Inversa**:
   Si deseamos generar una variable aleatoria $X$ con una Función de Distribución Acumulada (CDF) continua e invertible $F(x) = P(X \le x)$, podemos obtenerla transformando $U$:
   $$X = F^{-1}(U)$$
   *Ejemplo (Distribución Exponencial)*: La CDF es $F(x) = 1 - e^{-\lambda x}$ para $x \ge 0$.
   Su inversa es:
   $$U = 1 - e^{-\lambda x} \implies e^{-\lambda x} = 1 - U \implies x = -\frac{1}{\lambda} \ln(1 - U)$$
   Dado que $1-U$ se distribuye de la misma forma que $U$ en el intervalo $[0, 1)$, se suele simplificar como:
   $$X = -\frac{1}{\lambda} \ln(U)$$

2. **Método de Box-Muller (Distribución Normal)**:
   La CDF de la distribución normal no tiene una inversa en forma cerrada. Para generar muestras de una normal estándar $Z \sim \mathcal{N}(0, 1)$, se utiliza la transformación de Box-Muller, que toma dos variables independientes $U_1, U_2 \sim \mathcal{U}(0, 1)$ y genera un par de variables normales estándar independientes $Z_0, Z_1$:
   $$Z_0 = \sqrt{-2\ln(U_1)} \cos(2\pi U_2)$$
   $$Z_1 = \sqrt{-2\ln(U_1)} \sin(2\pi U_2)$$

## 4. Aplicaciones del Método de Montecarlo en la Industria

El Método de Montecarlo trasciende la teoría matemática pura, teniendo un impacto vital en múltiples sectores económicos y científicos modernos:

1. **Finanzas y Banca Cuantitativa**:
   * **Valoración de Opciones Exóticas**: Encontrar el valor justo de contratos financieros cuyas ganancias dependen de la trayectoria completa de los precios de los activos (como opciones asiáticas o de barrera), simulando miles de caminos aleatorios de precios bajo el Movimiento Browniano Geométrico.
   * **Análisis de Riesgo (Valor en Riesgo - VaR)**: Estimar las pérdidas máximas que un portafolio de inversión podría experimentar durante un período determinado bajo condiciones de mercado estocásticas.
2. **Logística y Transporte**:
   * **Simulación de Cadenas de Suministro**: Predecir cuellos de botella en la red de distribución mediante la modelación de los tiempos de aduanas, clima y accidentes viales.
   * **Redes de Colas en Aeropuertos y Terminales**: Planificar la cantidad de mostradores o andenes de carga requeridos analizando la llegada estocástica de clientes y el tiempo variable de servicio.
3. **Ciencias e Ingeniería**:
   * **Física y Blindaje Nuclear**: Simular la trayectoria libre media y colisiones de neutrones o fotones individuales al atravesar materiales, permitiendo diseñar reactores y escudos de radiación seguros.
   * **Confiabilidad Estructural**: Evaluar la probabilidad de falla de puentes, edificios o fuselajes de aviones, sometiéndolos a cargas de viento o sísmicas aleatorias bajo resistencia de materiales variable.
4. **Gestión de Proyectos**:
   * **Estimación de Plazos y Presupuestos**: Determinar la probabilidad de finalizar un proyecto complejo (por ejemplo, mediante el método PERT/CPM) antes de una fecha límite asignando distribuciones de probabilidad (como Beta o Triangular) a la duración de cada tarea crítica.
